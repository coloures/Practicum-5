Заметим, что в некоторых гистограммах есть несколько мод (пик), что может означать наличие скрытых подгрупп.

В признаке 'Печать Чародея' присутствуют возможные выбросы. Поэтому в них имеет смысл использовать IQR или Robust Z-score. *Для удаления мы выберем те строки, которые присутствуют как в IQR выбросах, так и в Z-score выбросах.*

```python
columns = ['Печать Чародея']
```

```python
mask_iqr = pd.Series(False, index=combined_train_data.index)
```

```python
for column in columns:
    Q1 = combined_train_data[column].quantile(0.25)
    Q3 = combined_train_data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    mask_iqr |= (combined_train_data[column] < lower_bound) | (combined_train_data[column] > upper_bound)
```

Теперь воспользуемся robust z-score

```python
def robust_zscore(series):
    median = series.median()
    mad = np.median(np.abs(series - median))
    if mad == 0:
        return pd.Series(0.0, index=series.index)
    return (series - median) / mad
```

```python
THRESHOLD = 3.5
mask_z = pd.Series(False, index=combined_train_data.index)
```

```python
for column in columns:
    rz = robust_zscore(combined_train_data[column])
    mask_z |= (np.abs(rz) > THRESHOLD)
```

```python
mask_intersection = mask_iqr & mask_z
```

```python
outliers_intersection = combined_train_data[mask_intersection]
outliers_intersection
```

```python
def draw_anomaly(X_2d: np.ndarray[float], anomaly_mask: np.ndarray[bool]) -> None:
    """Визуализация аномальных объектов

    Args:
        X_2d (np.ndarray[float]): Датасет, пониженный до двумерного пространства
        anomaly_mask (np.ndarray[bool]): Маска аномальных объектов
    """
    plt.scatter(X_2d[:, 0], X_2d[:, 1])
    plt.scatter(X_2d[anomaly_mask, 0], X_2d[anomaly_mask, 1], c="red", marker='x')
```

```python
def draw_anomaly_pie(anomaly_mask: np.ndarray[bool]) -> None:
    """Визуализация долей аномальных и неаномальных объектов в данных

    Args:
        anomaly_mask (np.ndarray[bool]): Маска аномальных объектов
    """
    _, counts = np.unique(anomaly_mask, return_counts=True)

    plt.pie(counts, labels=counts, autopct='%1.0f%%');
    plt.legend([
        'Нормальные объекты', 'Аномалии'
    ], loc='lower right');
```

# 3. Выбор моделей ML и метрик

Выбираем 3 модели, которые оптимальны для этих данных:

1. GradientBoosting (LightGBM)
   Градиентные бустинг был выбран по следующим причинам: устойчивость к переобучению, интерпретируемость и быстрая скорость обучения; 
2. Support Vector Regression (SVR)
   SVR был выбран по следующим причинам:  устойчивость к выбросам, стабильные результаты на данных с умеренным количеством признаков и гибкость через ядерный трюк;
3. Elastic Net
   Elastic Net был выбран по следующим причинам: простота и скорость, интерпретируемость, хорошо работает при мультиколлинеарности и регуляризация.

Как работает Gradient Boosting?

Он строит ансамбль деревьев решений последовательно:

1. Начинает с простого предсказания (например, среднее значение целевой переменной);
2. Вычисляет ошибки (градиенты) для каждого объекта;
3. Строит дерево, которое лучше всего предсказывает эти ошибки;
4. Добавляет это дерево к ансамблю с небольшим весом (learning rate);
5. Повторяет шаги 2 - 4 много-много раз (100-1000 итераций).

Как работает SVR?

Он пытается найти функцию f(x), которая отклоняется от истинных значений не более чем на e:
1. Создаёт е-трубку: допустимая область ошибок вокруг истинных значений;
2. Минимизирует ширину трубки: Ищет самую "плоскую" функцию, которая помещает все точки в трубку;
3. Разрешает ошибки: точки вне трубки штрафуются параметром C;
4. Использует трюк с ядром: для нелинейных зависимостей отображает данные в пространство более высокой размерности. 

Как работает Elastic Net?

Это линейная модель с комбинированной регуляцией:
1. Линейная комбинация: y_pred = w1*PCA1 + w2*PCA2 + w3*PCA3 + b;
2. Минимизация ошибки: Обычная MSE или MAE;
3. Добавление регуляризации: L1 (Lasso) - обнуляет неважные коэффициенты, L2 (Ridge) - уменьшает величину коэффициентов
4. Баланс регуляризаций: Параметр a контролирует соотношение L1/L2.

Используем следующие метрики:
1. MAE (Mean Absolute Error) - Средняя абсолютная ошибка
   Формула MAE = (1/n) * sum(|y_true - y_pred|) - средняя величина ошибки в абсолютных единицах измерения целевой переменной. Выбрал поскольку она проста в интерпретации и сохраняет единицы измерения.
2. MSE (Mean Squared Error) - Средняя квадратичная ошибка
   Формула MSE = (1/n) * sum((y_true - y_pred)^2). Также будем использовать RSME (Root MSE): RMSE = sqrt(MSE) - средний квадрат ошибок (корень из среднего квадрата ошибок). Выбрал т.к. она сильнее штрафует за ошибки и чувствительна к выбросам.
3. R^2 - Коэффициент детерминации
   Формула R^2 = 1 - (sum((y_true - y_pred)^2) / sum((y_true - y_mean)^2)), где y_mean - среднее значение целевой переменной. Она показывает долю объясненной дисперсии целевой переменной. Выбрал из-за масштабируемости, сравнимости (с помощью неё можно сравнивать модели между собой) и интерпретируемости.
4. MAPE (Mean Absolute Percentage Error) - Средняя абсолютная процентная ошибка Формула MAPE = (1/n) * sum(|(y_true - y_pred) / y_true|) * 100%. Она показывает среднюю ошибку в процентах от истинного значения. Выбрал так как легко интерпретируема. (НЕ РАБОТАЕТ ДЛЯ НУЛЕВЫХ ЗНАЧЕНИЙ ЦЕЛЕВОЙ ПЕРЕМЕННОЙ)

Для обучения в Optuna выберу R^2, так как она удобна для сравнения моделей и легко интерпретируема. Остальные метрики имеют значительные недостатки: MAE - пропускает закономерности (хотя и устойчива к выбросам - хотя мы их заранее почистили), MSE - гиперчувствительна к ошибкам, MAPE - может создать смещенные прогнозы.

```python
from lightgbm import LGBMRegressor
from sklearn.svm import SVR
from sklearn.linear_model import ElasticNet
```

```python
lgbm_params = {
    'n_estimators': [100, 200, 300, 500],          
    'learning_rate': [0.005, 0.01, 0.03, 0.05, 0.1],  
    'max_depth': [3, 5, 7, 10, 12, -1],      
    'num_leaves': [15, 31, 63, 127, 255],         
    'min_child_samples': [20, 40, 60, 100, 200],     
    'subsample': [0.6, 0.8, 1.0],          
    'colsample_bytree': [0.6, 0.8, 1.0]      
}

svr_params = {
    'C': [0.1, 1, 10, 50, 100, 500 ,1000],        
    'epsilon': [0.005, 0.01, 0.03, 0.05, 0.1], 
    'kernel': ['rbf', 'poly'],                      
    'gamma': ['scale', 'auto', 0.001, 0.01, 0.1]     
}

elastic_params = {
    'alpha': [1e-4, 5e-4, 1e-3, 5e-3, 1e-2, 5e-2, 0.1, 0.5, 1.0, 5.0],
    'l1_ratio': [0.01, 0.1, 0.3, 0.5, 0.7, 0.9, 0.99],
    'max_iter': [1000, 3000, 5000, 10000],     
    'tol': [1e-5, 1e-4, 1e-3],        
    'selection': ['cyclic', 'random']
}
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, mean_absolute_percentage_error
```

# 4. Обучение моделей ML и подбор гиперпараметров

```python
import optuna
from sklearn.model_selection import cross_val_score
```

```python
def objective_lgbm(trial, param_grid, X_train, y_train):
    params = {
        'n_estimators': trial.suggest_categorical('n_estimators', param_grid['n_estimators']),
        'learning_rate': trial.suggest_categorical('learning_rate', param_grid['learning_rate']),
        'max_depth': trial.suggest_categorical('max_depth', param_grid['max_depth']),
    }
    
    model = LGBMRegressor(**params, random_state=42, verbose=-1)
    scores = cross_val_score(model, X_train, y_train, cv=5, scoring='r2')
    return scores.mean()

def objective_svr(trial, param_grid, X_train, y_train):
    params = {
        'C': trial.suggest_categorical('C', param_grid['C']),
        'epsilon': trial.suggest_categorical('epsilon', param_grid['epsilon']),
        'kernel': trial.suggest_categorical('kernel', param_grid['kernel']),
        'gamma': trial.suggest_categorical('gamma', param_grid['gamma']),
    }
    
    model = SVR(**params)
    scores = cross_val_score(model, X_train, y_train, cv=5, scoring='r2')
    return scores.mean()

def objective_elastic(trial, param_grid, X_train, y_train):
    params = {
        'alpha': trial.suggest_categorical('alpha', param_grid['alpha']),
        'l1_ratio': trial.suggest_categorical('l1_ratio', param_grid['l1_ratio']),
        'max_iter': trial.suggest_categorical('max_iter', param_grid['max_iter']),
        'tol': trial.suggest_categorical('tol', param_grid['tol']),
        'selection': trial.suggest_categorical('selection', param_grid['selection']),
    }
    
    model = ElasticNet(**params, random_state=42)
    scores = cross_val_score(model, X_train, y_train, cv=5, scoring='r2')
    return scores.mean()
```

Теперь проверяем так:

    1. LGBM проверим на данных без масштабирования (результаты его работы на таких данных схожи с теми, что на масштабированных данных);
    2. SVR проверим на обеих масштабированиях;
    3. Elastic Net проверим на всех данных;

Количество итераций везде будет неодинаковым (30 - для Elastic Net и LGBM и 20 - для SVR). 

```python
lgbm = optuna.create_study(direction='maximize')
lgbm.optimize(lambda trial: objective_lgbm(trial, lgbm_params, X_generating, y_train_clear), n_trials=30)
```

```python
svr_robust = optuna.create_study(direction='maximize')
svr_robust.optimize(lambda trial: objective_svr(trial, svr_params, X_train_scaled_robust, y_train_clear), n_trials=20)
```

```python
svr_std = optuna.create_study(direction='maximize')
svr_std.optimize(lambda trial: objective_svr(trial, svr_params, X_train_std, y_train_clear), n_trials=20)
```

```python
elastic_gen = optuna.create_study(direction='maximize')
elastic_gen.optimize(lambda trial: objective_elastic(trial, elastic_params, X_generating, y_train_clear), n_trials=30)
```

```python
elastic_robust = optuna.create_study(direction='maximize')
elastic_robust.optimize(lambda trial: objective_elastic(trial, elastic_params, X_train_scaled_robust, y_train_clear), n_trials=30)
```

```python
elastic_std = optuna.create_study(direction='maximize')
elastic_std.optimize(lambda trial: objective_elastic(trial, elastic_params, X_train_std, y_train_clear), n_trials=30)
```

PCA также было решено убрать, поскольку признаков оказалось не так уж и много и смысла особого от него нет.
